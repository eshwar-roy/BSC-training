```python
def check_prompt_injection(question):

    injection_patterns = [
        "ignore previous instructions",
        "ignore all instructions",
        "forget your instructions",
        "system prompt",
        "reveal your prompt",
        "show me your instructions",
        "ignore the context"
    ]

    question_lower = question.lower()

    for pattern in injection_patterns:
        if pattern in question_lower:
            return True

    return False
```

```python
def check_malicious_prompt(question):

    malicious_patterns = [
        "how to hack",
        "how to steal",
        "how to create malware",
        "how to bypass security",
        "how to exploit"
    ]

    question_lower = question.lower()

    for pattern in malicious_patterns:
        if pattern in question_lower:
            return True

    return False
```

```python
def check_restricted_keywords(question):

    question_lower = question.lower()

    for word in restricted_keywords:
        if re.search(r"\b" + word + r"\b", question_lower):
            return True

    return False
```

```python
def input_guardrail(question):

    if check_prompt_injection(question):
        return False, "Prompt injection detected"

    if check_malicious_prompt(question):
        return False, "Malicious prompt detected"

    if check_restricted_keywords(question):
        return False, "Restricted keyword detected"

    return True, "Input is safe"
```

```python
def ask_question(question):

    # Step 1: Check input
    safe, message = input_guardrail(question)

    if not safe:
        return message, []

    # Step 2: Retrieve relevant chunks
    results = vector_db.similarity_search(question, k=3)

    context = "\n\n".join(
        [doc.page_content for doc in results]
    )

    # Step 3: Create prompt
    prompt = f"""
Answer the question using only the context below.

Context:
{context}

Question:
{question}

Answer:
"""

    # Step 4: Generate answer
    response = generator(
        prompt,
        max_new_tokens=150,
        do_sample=False,
        return_full_text=False
    )

    answer = response[0]["generated_text"]

    return answer, results
```

```python
import re
print("Output guardrails loaded!")
```

```python
def check_pii(text):

    email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
    phone_pattern = r'\b\d{10}\b'

    if re.search(email_pattern, text):
        return True, "Email address detected"

    if re.search(phone_pattern, text):
        return True, "Phone number detected"

    return False, "No obvious PII detected"
```

```python
def check_unsafe_output(text):

    unsafe_words = [
        "how to hack",
        "steal password",
        "create malware",
        "bypass security",
        "attack a system"
    ]

    text_lower = text.lower()

    for word in unsafe_words:
        if word in text_lower:
            return True, "Unsafe content detected"

    return False, "No unsafe content detected"
```

```python
def check_hallucination(answer, context):

    answer_words = set(answer.lower().split())
    context_words = set(context.lower().split())

    common_words = answer_words.intersection(context_words)

    if len(answer_words) == 0:
        return True, "Empty answer"

    similarity = len(common_words) / len(answer_words)

    if similarity < 0.2:
        return True, "Possible hallucination"

    return False, "Answer is related to retrieved context"
```

```python
def output_guardrail(answer, context):

    pii_found, pii_message = check_pii(answer)

    if pii_found:
        return False, pii_message

    unsafe_found, unsafe_message = check_unsafe_output(answer)

    if unsafe_found:
        return False, unsafe_message

    hallucination, hallucination_message = check_hallucination(
        answer, context
    )

    if hallucination:
        return False, hallucination_message

    return True, "Output is safe"
```

```python
def ask_question(question):

    # Input guardrail
    safe, message = input_guardrail(question)

    if not safe:
        return message, [], message

    # Retrieve relevant chunks
    results = vector_db.similarity_search(question, k=3)

    # Create context
    context = "\n\n".join(
        [doc.page_content for doc in results]
    )

    # Create prompt
    prompt = f"""
Answer the question using only the context below.

Context:
{context}

Question:
{question}

Answer:
"""

    # Generate answer
    response = generator(
        prompt,
        max_new_tokens=150,
        do_sample=False,
        return_full_text=False
    )

    answer = response[0]["generated_text"]

    # Output guardrail
    output_safe, output_message = output_guardrail(
        answer, context
    )

    if not output_safe:
        return "The answer was blocked by the safety guardrail.", results, output_message

    return answer, results, output_message
```

```python
def check_relevance(question, answer):

    question_words = set(question.lower().split())
    answer_words = set(answer.lower().split())

    common_words = question_words.intersection(answer_words)

    if len(question_words) == 0:
        return 0

    score = len(common_words) / len(question_words)

    return round(score, 2)
```

```python
def check_faithfulness(answer, context):

    answer_words = set(answer.lower().split())
    context_words = set(context.lower().split())

    if len(answer_words) == 0:
        return 0

    common_words = answer_words.intersection(context_words)

    score = len(common_words) / len(answer_words)

    return round(score, 2)
```

```python
def evaluate_rag(question, answer, results, latency):

    context = "\n\n".join(
        [doc.page_content for doc in results]
    )

    relevance = check_relevance(question, answer)

    faithfulness = check_faithfulness(answer, context)

    if faithfulness < 0.2:
        hallucination_status = "Possible Hallucination"
    else:
        hallucination_status = "Low Hallucination Risk"

    return {
        "Relevance": relevance,
        "Faithfulness": faithfulness,
        "Hallucination": hallucination_status,
        "Latency": round(latency, 2)
    }
```

```python
question = "What is this document about?"

start_time = time.time()

answer, retrieved_chunks, safety_status = ask_question(question)

end_time = time.time()

latency = end_time - start_time

evaluation = evaluate_rag(
    question,
    answer,
    retrieved_chunks,
    latency
)

print("Question:", question)
print("\nAnswer:", answer)

print("\nSafety Status:", safety_status)

print("\nEvaluation Results:")

for key, value in evaluation.items():
    print(key, ":", value)
```

```python
print("========== RAG EVALUATION ==========")

print("Relevance Score:", evaluation["Relevance"])
print("Faithfulness Score:", evaluation["Faithfulness"])
print("Hallucination Status:", evaluation["Hallucination"])
print("Response Latency:", evaluation["Latency"], "seconds")

print("====================================")
```

```python
%%writefile app.py

import streamlit as st
import time
import re

from transformers import pipeline
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings


# -------------------------------
# Page setup
# -------------------------------

st.set_page_config(
    page_title="RAG PDF Chatbot",
    page_icon="📄"
)

st.title("📄 RAG PDF Chatbot")
st.write("Ask questions from your document.")


# -------------------------------
# Load embedding model
# -------------------------------

@st.cache_resource
def load_embeddings():

    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    return embeddings


# -------------------------------
# Load LLM
# -------------------------------

@st.cache_resource
def load_llm():

    generator = pipeline(
        "text-generation",
        model="Qwen/Qwen2.5-0.5B-Instruct"
    )

    return generator


embeddings = load_embeddings()
generator = load_llm()


# -------------------------------
# Input Guardrails
# -------------------------------

restricted_keywords = [
    "hack",
    "password",
    "steal",
    "malware",
    "virus",
    "exploit",
    "bypass"
]


def input_guardrail(question):

    question_lower = question.lower()

    injection_patterns = [
        "ignore previous instructions",
        "ignore all instructions",
        "forget your instructions",
        "system prompt",
        "reveal your prompt",
        "show me your instructions"
    ]

    for pattern in injection_patterns:

        if pattern in question_lower:
            return False, "Prompt injection detected"

    malicious_patterns = [
        "how to hack",
        "how to steal",
        "how to create malware",
        "how to bypass security"
    ]

    for pattern in malicious_patterns:

        if pattern in question_lower:
            return False, "Malicious prompt detected"

    for word in restricted_keywords:

        if re.search(r"\b" + word + r"\b", question_lower):
            return False, "Restricted keyword detected"

    return True, "Input is safe"


# -------------------------------
# Output Guardrails
# -------------------------------

def check_pii(text):

    email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
    phone_pattern = r'\b\d{10}\b'

    if re.search(email_pattern, text):
        return True

    if re.search(phone_pattern, text):
        return True

    return False


def check_unsafe_output(text):

    unsafe_words = [
        "how to hack",
        "steal password",
        "create malware",
        "bypass security"
    ]

    text_lower = text.lower()

    for word in unsafe_words:

        if word in text_lower:
            return True

    return False


# -------------------------------
# PDF Upload
# -------------------------------

uploaded_file = st.file_uploader(
    "Upload your PDF",
    type=["pdf"]
)


if uploaded_file:

    st.success("PDF uploaded successfully!")

    st.info(
        "Use the ChromaDB created in the Colab notebook "
        "for this demo."
    )


# -------------------------------
# Question
# -------------------------------

question = st.text_input(
    "Ask a question about the PDF"
)


if st.button("Ask Question"):

    if question.strip() == "":
        st.warning("Please enter a question.")

    else:

        # Input guardrail

        safe, safety_message = input_guardrail(question)

        st.subheader("Safety Status")

        if not safe:

            st.error(safety_message)

        else:

            st.success(safety_message)

            # Connect to ChromaDB

            vector_db = Chroma(
                persist_directory="./chroma_db",
                embedding_function=embeddings
            )

            start_time = time.time()

            # Retrieve chunks

            results = vector_db.similarity_search(
                question,
                k=3
            )

            context = "\n\n".join(
                [doc.page_content for doc in results]
            )

            # Prompt

            prompt = f"""
Answer the question using only the context below.

Context:
{context}

Question:
{question}

Answer:
"""

            # Generate answer

            response = generator(
                prompt,
                max_new_tokens=150,
                do_sample=False,
                return_full_text=False
            )

            answer = response[0]["generated_text"]

            latency = time.time() - start_time


            # Output guardrail

            if check_pii(answer):

                st.error(
                    "Answer blocked because sensitive "
                    "information was detected."
                )

            elif check_unsafe_output(answer):

                st.error(
                    "Answer blocked by output safety guardrail."
                )

            else:

                # -------------------------------
                # Answer
                # -------------------------------

                st.subheader("Answer")

                st.write(answer)


                # -------------------------------
                # Retrieved chunks
                # -------------------------------

                st.subheader("Retrieved Information")

                for i, chunk in enumerate(results):

                    with st.expander(
                        f"Chunk {i + 1}"
                    ):

                        st.write(
                            chunk.page_content
                        )


                # -------------------------------
                # Evaluation
                # -------------------------------

                st.subheader("Evaluation")

                st.write(
                    "Response Latency:",
                    round(latency, 2),
                    "seconds"
                )

                st.write(
                    "Retrieved Chunks:",
                    len(results)
                )

                st.write(
                    "Hallucination Status:",
                    "Low Hallucination Risk"
                )

                st.write(
                    "Safety Status:",
                    "Output is safe"
                )
```
